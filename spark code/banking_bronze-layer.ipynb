```python


##Create a table called for customers

customers_bronze = spark.read.csv(
    "/Volumes/banking-proj/default/banking_data/customers.csv",
    header=True,
    inferSchema=True

)

display(customers_bronze)
```

```python
## create table for accounts

accounts_bronze = spark.read.csv(
    "/Volumes/banking-proj/default/banking_data/accounts.csv",
    header=True,
    inferSchema=True)

display(accounts_bronze)
```

```python
##create table for logins

logins_bronze = spark.read.csv("/Volumes/banking-proj/default/banking_data/logins.csv",
            header=True,
            inferSchema=True)

display(logins_bronze)
```

```python
##creating table for transactions

transactions_bronze = spark.read.csv("/Volumes/banking-proj/default/banking_data/transactions.csv",
            header=True,
            inferSchema=True)

display(transactions_bronze)
```

```python
## creating table from df.

customers_bronze.write.mode("overwrite").saveAsTable("customers_bronze_tb")

accounts_bronze.write.mode("overwrite").saveAsTable("accounts_bronze_tb")

logins_bronze.write.mode("overwrite").saveAsTable("logins_bronze_tb")

transactions_bronze.write.mode("overwrite").saveAsTable("transactions_bronze_tb")
```
