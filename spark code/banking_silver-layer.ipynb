```python
customers_silver = spark.table("customers_bronze_tb")
accounts_silver = spark.table("accounts_bronze_tb")
logins_silver = spark.table("logins_bronze_tb")
transactions_silver = spark.table("transactions_bronze_tb")
```

```python
##check dups

##table count
customers_silver.count()

##after removing dups what is the table count
customers_silver.dropDuplicates().count()
```

```python
## clean duplicates

customers_clean = customers_silver.dropDuplicates()
customers_clean.count()
```

```python
# check for null values

from pyspark.sql.functions import *

null_columns = []

for c in customers_clean.columns:
    null_count = count(
    when 
        (col(c).isNull(),
         c
        )
    ).alias(c)
    
    null_columns.append(null_count)

customers_clean.select(null_columns).show()
```

```python
## Clean dups in transactions table

transactions_clean = transactions_silver.dropDuplicates()
transactions_clean.count()
```

```python
## change datatype from string to timestamp

from pyspark.sql.functions import *

transactions_clean = transactions_clean.withColumn("timestamp", to_timestamp(col("timestamp")))
```

```python
## Add audit column

transactions_clean = transactions_clean.withColumn("load_time", current_timestamp())
```

```python
##Save Silver tables

customers_clean.write.mode("overwrite").saveAsTable("silver_customers")
transactions_clean.write.mode("overwrite").saveAsTable("silver_transactions")
logins_silver.write.mode("overwrite").saveAsTable("silver_logins")
accounts_silver.write.mode("overwrite").saveAsTable("silver_accounts")
```
